```python
dinheiro = float(input("Digite o valor do seu dinheiro: "))
iphone = 1500
saldo = dinheiro - iphone

if (dinheiro >= iphone):
  print(f"Pode comprar a iphone, seu saldo atual é: {saldo} Reais.")
else:
  print("Dinheiro insuficiente")
```

```python
diametro = float(input("Digite o diâmetro do círculo: "))
R = diametro / 2
area = R * R * 3.14
print(f"A área do Circulo é: {area}")
```

```python
B = int(input("Digite o valor da base do quadrado: "))
H = int(input("Digite o valor da altura do quadrado: "))
area = B * H
print(f"A área do quadrado é: {area}")
```

```python
b = int(input("Digite o valor de b: "))
c = int(input("Digite o valor de c: "))
catetoOposto = b ** 2
catetoAdjacente = c ** 2
somaDosQuadrados = catetoOposto + catetoAdjacente
hipotenusa = somaDosQuadrados ** 0.5
print(f"A hipotenusa é: {hipotenusa}")
```
